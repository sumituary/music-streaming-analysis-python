```python
import pandas as pd
import matplotlib.pyplot as plt

#1. MUSIC STREAMING DASHBOARD


# 2. CREATING DATASET
#__________________________________

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "SongsPlayed": [1200, 1450, 1700, 1600, 1900, 2200],
    "ActiveUsers": [500, 550, 620, 610, 700, 780],
    "PremiumSubscribers": [180, 200, 240, 230, 270, 310],
    "ListeningHours": [3200, 3500, 3900, 3700, 4200, 4600],
    "NewPlaylists": [90, 110, 130, 120, 150, 180],
    "NewUsers": [120, 140, 160, 150, 190, 220],
    "Revenue": [1800, 2100, 2500, 2400, 2900, 3400],
    "Likes": [800, 950, 1100, 1050, 1250, 1450]
}

# Convert dictionary into DataFrame
df = pd.DataFrame(data)
df      


# 3. CREATE DASHBOARD
#__________________________________

# Create 4 rows × 2 columns
fig, axes = plt.subplots(4, 2, figsize=(14, 16))

# Main title
fig.suptitle(
    "Music Streaming Analytics Dashboard",
    fontsize=20,
    fontweight="bold"
)



# 4. CHART 1 → SONGS PLAYED
#__________________________________

df.plot(
    x="Month",
    y="SongsPlayed",
    ax=axes[0, 0],
    title="Songs Played Trend",
    color="royalblue",
    marker="o"
)

axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Songs Played")
axes[0, 0].grid(axis="y", alpha=0.3)


# 5. CHART 2 → ACTIVE USERS
#__________________________________

df.plot(
    x="Month",
    y="ActiveUsers",
    kind="bar",
    ax=axes[0, 1],
    title="Monthly Active Users",
    color="mediumseagreen"
)

axes[0, 1].set_xlabel("Month")
axes[0, 1].set_ylabel("Users")
axes[0, 1].grid(axis="y", alpha=0.3)



# 6. CHART 3 → NEW USERS
#__________________________________

df.plot(
    x="Month",
    y="NewUsers",
    kind="area",
    ax=axes[1, 0],
    title="New Users Growth",
    color="mediumorchid",
    alpha=0.6
)

axes[1, 0].set_xlabel("Month")
axes[1, 0].set_ylabel("New Users")
axes[1, 0].grid(axis="y", alpha=0.3)



# 7. CHART 4 → LISTENING HOURS
#__________________________________

df.plot(
    x="Month",
    y="ListeningHours",
    ax=axes[1, 1],
    title="Listening Hours Trend",
    color="darkorange",
    marker="o"
)

axes[1, 1].set_xlabel("Month")
axes[1, 1].set_ylabel("Hours")
axes[1, 1].grid(axis="y", alpha=0.3)


# 8. CHART 5 → REVENUE
#__________________________________

df.plot(
    x="Month",
    y="Revenue",
    kind="bar",
    ax=axes[2, 0],
    title="Monthly Revenue",
    color="crimson"
)

axes[2, 0].set_xlabel("Month")
axes[2, 0].set_ylabel("Revenue ($)")
axes[2, 0].grid(axis="y", alpha=0.3)



# 9. CHART 6 → LIKES
#__________________________________

df.plot(
    x="Month",
    y="Likes",
    ax=axes[2, 1],
    title="Likes Trend",
    color="teal",
    marker="o"
)

axes[2, 1].set_xlabel("Month")
axes[2, 1].set_ylabel("Likes")
axes[2, 1].grid(axis="y", alpha=0.3)



# 10. CHART 7 → ACTIVE USERS VS PREMIUM USERS
#__________________________________

df.plot(
    x="Month",
    y=["ActiveUsers", "PremiumSubscribers"],
    kind="bar",
    ax=axes[3, 0],
    title="Active Users vs Premium Users",
    color=["mediumseagreen", "mediumorchid"]
)

axes[3, 0].set_xlabel("Month")
axes[3, 0].set_ylabel("Users")
axes[3, 0].grid(axis="y", alpha=0.3)


# 11. CHART 8 → SONGS PLAYED VS LIKES
#__________________________________

axes[3, 1].scatter(
    df["SongsPlayed"],
    df["Likes"],
    color="gold",
    edgecolors="black",
    s=100
)

axes[3, 1].set_title("Songs Played vs Likes")
axes[3, 1].set_xlabel("Songs Played")
axes[3, 1].set_ylabel("Likes")
axes[3, 1].grid(alpha=0.3)



# 12. DASHBOARD FORMATTING
#__________________________________

plt.tight_layout(
    rect=[0, 0, 1, 0.96]
)

plt.show()



# 13. BASIC DATA ANALYSIS
#__________________________________

print("----- Basic Data Analysis -----")

print(
    "Total Songs Played:",
    df["SongsPlayed"].sum()
)

print(
    "Average Active Users:",
    df["ActiveUsers"].mean()
)

print(
    "Total Listening Hours:",
    df["ListeningHours"].sum()
)

print(
    "Total Revenue:",
    df["Revenue"].sum()
)

print(
    "Average Likes:",
    df["Likes"].mean()
)



# 14. HIGHEST VALUES
#__________________________________

print("\n----- Highest Values -----")

index = df["SongsPlayed"].idxmax()

print(
    "Highest Songs Played:",
    df.loc[index, "Month"],
    "(",
    df.loc[index, "SongsPlayed"],
    ")"
)


index = df["ActiveUsers"].idxmax()

print(
    "Highest Active Users:",
    df.loc[index, "Month"],
    "(",
    df.loc[index, "ActiveUsers"],
    ")"
)


index = df["Revenue"].idxmax()

print(
    "Highest Revenue:",
    df.loc[index, "Month"],
    "($",
    df.loc[index, "Revenue"],
    ")"
)


index = df["ListeningHours"].idxmax()

print(
    "Highest Listening Hours:",
    df.loc[index, "Month"],
    "(",
    df.loc[index, "ListeningHours"],
    "hours)"
)


# 15. KEY INSIGHTS
#__________________________________

print("\n----- Key Insights -----")

print(
    "1. Songs played increased from January to June."
)

print(
    "2. Active users increased over the six-month period."
)

print(
    "3. New users generally increased over time."
)

print(
    "4. Listening hours increased as the platform grew."
)

print(
    "5. Revenue was highest in June."
)

print(
    "6. Likes generally increased with songs played."
)

print(
    "7. Premium subscribers increased along with active users."
)

print(
    "8. Overall, the music streaming platform shows positive growth."
)
```
